```python
### 10-15-25 DSAI201

def find_deriv(f, a, h):
    print(f"Finding derivative of {f} with a={a} and h={h}")
    return (f(a+h) - f(a)) / h


def main():
    def f1(x):
        return x**2 # simple polynomial
    
    def f2(x):
        return 4*x**3 + 1/3*x**2 - 23*x + 2 # random polynomial


    # test settings
    a = 3
    

    h_list = [-0.01, -0.001, 0.001, 0.01]
    results = [None] * len(h_list)

    
    print(f"Testing: f1: with a={a}")
    for i, h in enumerate(h_list):
        results[i] = find_deriv(f1, a, h)
        print(f"Deriv of f1({a}) with h={h}: {results[i]}")

    print(results)


if __name__ == "__main__":
    main()
```
