## A live update for variance
For this problem, we will use the variance formula with delta degrees of freedom = 0,
\begin{equation}
    \text{variance} = \frac{1}{N} \sum_{j=1}^{N}(x_j - \bar{x})^2,
\end{equation}
which has the equivalent formulation (can you verify it?)
\begin{equation}
    \text{variance} = \frac{1}{N}\sum_{j=1}^N x_j^2 - \left(\frac{1}{N}\sum_{j=1}^N x_j\right)^2
\end{equation}
