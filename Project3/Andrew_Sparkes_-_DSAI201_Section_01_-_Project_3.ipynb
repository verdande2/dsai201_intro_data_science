## Adding a column for batting average
A player's batting average is computed by taking the number of hits divided by the number of at bats,
\begin{equation}
\text{batting average} = \frac{\text{\# Hits}}{\text{\# At Bats}}.
\end{equation}
These statistics for each player are found under the `H` and `AB` columns, respectively.

**PROMPT** Add a column to `hitting` called `avg` that provides the players batting average. For players that played on multiple teams, compute their average while playing for each team (so two different averages for two different teams).

## A live update for variance
For this problem, we will use the variance formula with delta degrees of freedom = 0,
\begin{equation}
    \text{variance} = \frac{1}{N} \sum_{j=1}^{N}(x_j - \bar{x})^2,
\end{equation}
which has the equivalent formulation (can you verify it?)
\begin{equation}
    \text{variance} = \frac{1}{N}\sum_{j=1}^N x_j^2 - \left(\frac{1}{N}\sum_{j=1}^N x_j\right)^2
\end{equation}

First, some definitions to use:
$$ \bar X = N^{-1} \sum_{j = 1} ^ N X_j $$
$$ \bar X_N = N^{-1}[X_N + (N-1)\bar X_{N-1}] $$
$$ s^2_N = \frac{\sum_{j = 1} ^ N (X_j - \bar X_N)^2}{N - 1}
= \frac{\sum_{j = 1} ^ N (X_j - \bar X_{N - 1} + \bar X_{N - 1} - \bar X_N)^2}{N - 1} $$
$$ s^2_N = \frac{(N - 2)s^2_{N - 1} + (N - 1)(\bar X_{N - 1} - \bar X_N)^2
+ (X_N - \bar X_{N})^2}{N - 1} $$
$$ s_N^2={N-2\over N-1}s^2_{N-1}+{1\over N}(X_N-\bar X_{N-1})^2 $$


Equation 1 shows the definition for our sample mean $ \bar{X} $. Equation 2 is the incremental/iterative version of that same sample mean, after adding one new value. Equation 3 shows the beginning steps of defining the variance in terms of totals and sums, simplifying, some symbol manipulation, the summation goes to zero, and Equation 5 is the simplified result for calculating the incremental/iterative variance after adding a new data point.
